```python
''' A empresa Hipotheticus paga R$10,00 por hora normal trabalhada, e R$15,00
por hora extra. Faça um algoritmo para calcular e imprimir o salário bruto e o
salário líquido de um determinado funcionário. Considere que o salário líquido é
igual ao salário bruto descontando-se 10% de impostos. ''' 

hora_normal = 10
hora_extra = 15
horas_trabalhadasN = float(input('digite quantas horas você trabalhou'))
horas_trabalhadasE = float(input('digite quantas horas extras você trabalhou'))

salario_bruto = (horas_trabalhadasN * hora_normal) + (horas_trabalhadasE * hora_extra)
desconto = salario_bruto * 10 / 100
salario_liquido = salario_bruto - desconto 

print(f'o salario bruto é {salario_bruto}, o salario liquido, com o desconto de {desconto}, sera de {salario_liquido}')
```
